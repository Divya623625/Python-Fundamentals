Pandas is a Python library used to store, organize, clean, manipulate and analyze data. It is mainly used for working with tables containing rows and columns.

Why do we use Pandas?
- To Read CSV and Excel files.
- To clean missing values.
- To remove duplicate records.
- To filter and sort data.
- To perform calculations.
- To analyze large datasets.
- To prepare data for machine learning.

```python
# AQI Data set
df = pd.read_csv("globalAirQuality.csv")
print(df)
```

```python
# selecting data
print(df["city"]) # selecting one column
print(df[["city","aqi"]]) # selectig multiple columns
print(df.loc[1]) # selects data rows using loc[]
print(df.loc[0:2]) # selects multiple rows - start idx : end idx (inclusive)
print(df.iloc[0]) # selects data using integer positions
print(df.iloc[0:2]) # selects first two rows - start idx : end idc (exclusive)
print(df.loc[0,["aqi","city"]]) # selecting a paticular value
print(df.iloc[0:2,0:2]) # selects row and columns together
print(df.iloc[0:2,2])

# cell
print(df.at[0,"city"])
print(df.iat[0, 2])

cities = df["city"].copy()
cities[0] = "Mumbai" # copy
```

```python
# Filtering of data
print(df[df["aqi"] > 100])
print(df[(df["aqi"] > 100) & (df["temperature"] > 30)])
print(df[(df["aqi"] > 100) & (df["temperature"] < 30)])
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] < 30)][["city","aqi"]]
print(aqi_data.iloc[0])

# Query
aqi_val = 100
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]
df.query("aqi > @aqi_val & temperature > 30")[["city","aqi"]].copy()
print(aqi_data)
```

```python
# Cleaning data - Handling missing values
df = pd.read_csv("raw_data.csv")
print(df)
print(df.isnull())
print(df.isnull().sum())
print(df.fillna(0))

age_mean = df["age"].mean()
cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
print(cleaned_data)
df.ffill()
df.bfill()
```

```python
# handle duplicates
df[["country", "gender"]].duplicated()
df2 = df.copy()

print(df.drop_duplicates(inplace = True))
```

```python
# Handle data types
df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64").copy()
df2.dtypes

date_str = pd.Series([pd.to_datetime("2026-12-31")])
type(date_str.dtypes)
```

```python
df["gender"].str.capitalize()
df["name"].str.split(" ")
df["country"].str.contains("india", case=False)
```

```python
# Feature engineering
# apply() - applies a function to a series or a dataframe
df2 = df.copy()
df2["tax"] = df2["income"].apply(lambda x: "20%" if x >= 6000 else "10%")
print(df2)

# map() - maps values in a series
gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df2["gender"] = df2["gender"].map(gender_map)
print(df2)

# assign
df2 = df2.assign(new_income = df2["income"] * 1.1)
print(df2)

# replace(old, new)
df2["country"] = df2["country"].replace("USA","US")
print(df2)

#rename(columns={old:new})
df2.columns = ["Id","Name","Age","Country","Gender","Income","Tax","New_Income"]
df2 = df2.rename(columns={"Income":"Salary"})
df2 = df2.rename(index={1:"First"})
print(df2)

# sort_values
df2 = df2.fillna(50)
sorted_df = df2.sort_values(["Salary", "Age"], ascending = True)
print(sorted_df)

# sort index
df2.sort_index(ascending = False)

# reset index
sorted_df = sorted_df.reset_index(drop=True)
print(sorted_df)

# Ranking 
sorted_df["Ranking"] = sorted_df["Salary"].rank(ascending=False, method="dense")
print(sorted_df)
```

```python
import pandas as pd
df = pd.read_csv("IRIS.csv")
```

```python
# First 10 rows
print(df.head(10))

# Shape and data types
print(df.shape)
print(df.dtypes)

# Summary statistics
print(df.describe())

# Selecting the rows where petal_length > 4.5 and species = virginica
print(df[(df['petal_length'] > 4.5)  & (df['species'] == 'virginica')])

# Group by species
print(df.groupby('species').agg({
    'sepal_length':'mean',
    'petal_length':'max',
    'sepal_length':'std'
}))

# Create petal_ratio and find its average by species
df['petal_ratio'] = df['petal_length'] / df['petal_width']
print('petal_ratio')

print(df.groupby('species')['petal_ratio'].mean())
```
